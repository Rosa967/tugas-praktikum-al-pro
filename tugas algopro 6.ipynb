```python
# a. Membuat Variabel Global
# Membuat variabel global skor_nobita = 0 untuk menyimpan skor nobita selama simulasi berlangsung.
skor_nobita = 0

# b. Membuat Fungsi untuk Memperbarui Skor
# Buat fungsi update_skor(nilai) yang akan menambah nilai ke skor_nobita. 
# Gunakan kata kunci global agar fungsi bisa mengubah variabel global di luar fungsi.
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai

# c. Membuat Fungsi Aktivitas Nobita

# 1. Fungsi belajar(jam_belajar)
# Gunakan logika if-elif-else untuk menentukan skor berdasarkan jam belajar dan gunakan return untuk mengembalikan nilai skor.
def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 50
    elif 1 <= jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:  # jam_belajar >= 3
        print("Nobita belajar dengan rajin hari ini")
        return 10

# 2. Fungsi tidur (jam_tidur)
# Gunakan if-elif-else untuk mengatur efek waktu tidur dan gunakan return untuk mengembalikan nilai skor.
def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif 1 <= jam_tidur <= 2:
        print("Waktu tidur Nobita seimbang")
        return 20
    else:  # jam_tidur > 3 (atau kondisi lainnya)
        print("Nobita tidur terlalu lama")
        return -20

# 3. Fungsi main_bersama_doraemon(jumlah_gadget, waktu_main)
# Gunakan if-elif-else untuk mengatur jumlah gadget dan waktu main dan gunakan return untuk mengembalikan nilai skor.
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain")
        return -30

# 4. Fungsi menyontek()
# Fungsi ini tidak membutuhkan parameter, namun tetap memiliki nilai kembalian (return) skor. 
# Berikan output "Nobita ketahuan menyontek" dan berikan skor (-50).
def menyontek():
    print("Nobita ketahuan menyontek")
    return -50
```

```python
# Contoh simulasi aktivitas Nobita:
print(f"Skor awal: {skor_nobita}")

# Nobita belajar selama 2 jam
skor_aktivitas = belajar(2)
update_skor(skor_aktivitas)

# Nobita ketahuan menyontek
skor_aktivitas = menyontek()
update_skor(skor_aktivitas)

print(f"Skor akhir: {skor_nobita}")
```
